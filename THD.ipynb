### 1. Título del proyecto

Música y salud mental: análisis exploratorio de la relación entre los hábitos de escucha, las preferencias musicales y los indicadores de bienestar psicológico.

**Objetivo general**

Explorar posibles asociaciones entre los hábitos de consumo musical, las preferencias por determinados géneros y los indicadores de salud mental autodeclarados por los participantes, concretamente ansiedad, depresión, insomnio y síntomas de TOC.

### 2. Hipótesis de trabajo
Código	Hipótesis	Variables principales

- H1	Las horas de escucha musical están asociadas con los niveles de ansiedad.	Hours per day / Anxiety

- H2	Los niveles de depresión presentan diferencias según el género musical favorito.	Fav genre / Depression

- H3	La frecuencia de escucha de determinados géneros está asociada con los indicadores psicológicos.	Frequency / Anxiety, Depression, Insomnia, OCD

- H4	Existen diferencias en los indicadores psicológicos según la percepción de los participantes sobre los efectos de la música.	Music effects / indicadores psicológicos

Estas hipótesis son preguntas que vamos a contrastar, no resultados que demos por demostrados.
