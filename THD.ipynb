## Análisis Empírico de la Disonancia Acústico-Emocional y su Impacto en la Soberanía Atencional

### 1. Enfoque y Propósito
El proyecto investiga el impacto invisible que la industria de consumo musical masivo ejerce sobre el subconsciente colectivo y la salud mental. Partiendo de los principios de condicionamiento y repetición, la investigación busca demostrar empíricamente cómo el consumo sostenido de narrativas de conflicto y vacío emocional actúa como un factor de fricción en procesos de bienestar personal.


### 2. Hipótesis de Investigación

*   **Hipótesis 1: El Sesgo de Valencia en el Consumo Masivo**
    *   *Planteamiento:* ¿El ecosistema de reproducción comercial global prioriza sistemáticamente estímulos ligados a la tensión afectiva, el conflicto o el vacío emocional frente a la estabilidad?.
    *   *Métrica:* Correlación directa entre la popularidad global (`popularity`) y la valencia emocional (`valence`).
*   **Hipótesis 2: La Disonancia Acústico-Emocional en la Música Urbana**
    *   *Planteamiento:* ¿La música urbana comercial genera disonancia cognitiva al fusionar un estímulo físico hiperactivo (alta energía motora) con un relato lírico de pesimismo o conflicto afectivo?.
    *   *Métrica:* Concentración de registros en el cuadrante de **Alta Energía (`energy`) y Baja Valencia (`valence`)**, segmentado por géneros (`track_genre`).

## 4. Consideraciones Metodológicas y Alcance
* **Variables Clave:** El estudio se centrará en las métricas acústicas y de impacto (`popularity`, `valence`, `energy`, `danceability`) cruzadas con la categorización por género (`track_genre`).
* **Control de Duplicados por Género:** Dado que en este catálogo de 114.000 registros una misma canción puede aparecer etiquetada bajo múltiples subgéneros, se aplicara el identificador único (`track_id`) para evitar sesgos de duplicidad al calcular promedios globales.
* **Escalabilidad Visual:** Al trabajar con 125 géneros, el análisis visual se apoyará en segmentaciones estratégicas (comparativa entre Top géneros comerciales vs. géneros de nicho) para asegurar gráficos limpios, legibles y directos.
* **Tipología de Visualizaciones (`Seaborn` / `Matplotlib`):** (Susceptible de cambios en función de si se entienden o las hipótesis varían o aparecen más)
  * **Gráficos de dispersión (`scatterplot` / `jointplot`):** Para cruzar `energy` vs. `valence` y delimitar visualmente los cuatro cuadrantes emocionales, aislando la zona de disonancia acústico-emocional.
  * **Diagramas de caja y violín (`boxplot` / `violinplot`):** Para comparar la distribución de la positividad (`valence`) y la bailabilidad (`danceability`) entre los géneros comerciales de alta rotación frente al resto del catálogo.
  * **Mapas de calor (`heatmap`):** Para cuantificar y visualizar de un vistazo las matrices de correlación entre popularidad y métricas acústicas.
  * **Gráficos de barras comparativos (`barplot`):** Para rankear los géneros con mayor brecha entre energía física y valencia emocional.
