# Primer análisis de variables categóricas

```python
# Distribución del género musical favorito
df["Fav genre"].value_counts()
```

```python
# Distribución del género musical favorito

generos = df["Fav genre"].value_counts()

distribucion_generos = pd.DataFrame({
    "Participantes": generos,
    "Porcentaje": (generos / len(df) * 100).round(2)
})

distribucion_generos
```

### Distribución de los géneros musicales favoritos

El análisis de las preferencias musicales muestra que Rock es el género favorito más frecuente, con 188 participantes (25,54 %), seguido de Pop, con 114 participantes (15,49 %), y Metal, con 88 participantes (11,96 %).

Estos tres géneros concentran aproximadamente el 52,99 % de los participantes, lo que refleja una distribución desigual de las preferencias musicales.

Por otro lado, géneros como Latin y Gospel presentan una representación reducida dentro de la muestra.

Esta distribución deberá tenerse en cuenta en el análisis posterior de la relación entre género musical favorito y niveles de depresión, especialmente por las diferencias en el tamaño de los grupos.

```python
# Distribución de la percepción de los efectos de la música

efectos = df["Music effects"].value_counts()

distribucion_efectos = pd.DataFrame({
    "Participantes": efectos,
    "Porcentaje": (
        efectos / df["Music effects"].notna().sum() * 100
    ).round(2)
})

distribucion_efectos
```

```python
# Comprobar los valores nulos de esta variable

print("Total de participantes:", len(df))
print("Respuestas válidas:", df["Music effects"].notna().sum())
print("Respuestas nulos:", df["Music effects"].isna().sum())
```

### Distribución de la percepción de los efectos de la música

La variable Music effects recoge la percepción de los participantes sobre cómo influye la música en su salud mental.

De los 736 participantes, 728 proporcionaron una respuesta válida y 8 registros presentan valores ausentes.

Entre las respuestas válidas, 542 participantes (74,45 %) consideran que la música mejora su salud mental, 169 (23,21 %) indican que no tiene efecto y 17 (2,34 %) manifiestan que la empeora.

Se observa un claro predominio de la categoría Improve. Asimismo, la categoría Worsen presenta una representación muy reducida, circunstancia que deberá tenerse en cuenta al realizar las comparaciones estadísticas correspondientes a la hipótesis H4.

Estos resultados representan percepciones subjetivas y no permiten establecer por sí solos una relación causal entre la música y la salud mental.
