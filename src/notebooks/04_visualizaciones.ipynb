```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/spotify_procesado01.csv")

resumen = df.groupby("track_genre").agg({
    "energy": "mean",
    "valence": "mean",
    "popularity": "mean"
}).reset_index()

plt.figure(figsize=(10, 6))
plt.scatter(resumen["energy"], resumen["valence"], c=resumen["popularity"], cmap="viridis", alpha=0.7, s=50)
plt.colorbar(label="Popularidad Media")
plt.title("Mapa Acústico Global: Energía vs Valencia por Género")
plt.xlabel("Energía Media")
plt.ylabel("Valencia Media (Positividad)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.savefig("../data/mapa_acustico.png", dpi=300, bbox_inches="tight")
plt.show()
```

Dispersión y Densidad: La gran mayoría de los géneros se concentran en una franja de energía media-alta (0.5 a 0.8) pero con una valencia moderada-baja, lo que demuestra que la música de Spotify tiende de forma natural al dinamismo rítmico combinado con tonalidades introspectivas o tensas.El rol del color (Popularidad): Los puntos con mayor popularidad (los tonos más brillantes de la paleta de color) no se limitan a la esquina superior derecha (alta energía y alta alegría/valencia), sino que se distribuyen hacia zonas de tensión emocional y alta bailabilidad, validando que el mercado global recompensa el estímulo rítmico y la carga emocional por encima de la "alegría" plana.
