# Música y salud mental: análisis exploratorio de datos

## 1. Introducción y objetivos

La música forma parte de nuestra vida cotidiana y muchas personas la utilizan para relajarse, concentrarse o gestionar sus emociones.

En este proyecto se realiza un análisis exploratorio de datos (EDA) para investigar posibles relaciones entre los hábitos de escucha musical, las preferencias musicales y diferentes indicadores de salud mental.

El objetivo principal es aplicar los conocimientos adquiridos durante el módulo de Data Analysis, utilizando Python y librerías como Pandas, Matplotlib y Seaborn para limpiar, analizar y visualizar datos.

Para orientar el análisis se plantean cuatro preguntas:

1. ¿Existe alguna relación visual entre las horas diarias de escucha musical y los niveles de ansiedad?
2. ¿Se observan diferencias en los niveles de depresión según el género musical favorito?
3. ¿Cómo varían los indicadores de salud mental según la frecuencia de escucha de distintos géneros musicales?
4. ¿Existen diferencias en los indicadores psicológicos según los efectos que los participantes atribuyen a la música?

El propósito no es demostrar relaciones causales, sino identificar patrones y diferencias descriptivas dentro del conjunto de datos.

## 2. Descripción del conjunto de datos

Para realizar el proyecto se ha utilizado el dataset **Music & Mental Health Survey (MxMH)**, disponible en Kaggle.

El conjunto de datos contiene 736 respuestas y 33 columnas, con información sobre los hábitos de escucha musical y la salud mental autodeclarada de los participantes.

Entre las variables principales encontramos:

- **Age:** edad del participante.
- **Hours per day:** horas diarias de escucha musical.
- **Fav genre:** género musical favorito.
- **Frequency [género]:** frecuencia de escucha de 16 géneros musicales.
- **Anxiety, Depression, Insomnia y OCD:** puntuaciones autodeclaradas entre 0 y 10.
- **Music effects:** percepción de los efectos de la música sobre la salud mental.

Los datos proceden de una encuesta y reflejan las respuestas de sus participantes, por lo que no deben interpretarse como diagnósticos médicos ni como una muestra necesariamente representativa de toda la población.

## 3. Exploración y preparación de los datos

Inicialmente se realizó una exploración del dataset para conocer su estructura, identificar valores nulos, comprobar posibles duplicados y revisar las distribuciones de las variables.

Durante este proceso se observaron los siguientes aspectos:

- El dataset original contiene 736 registros y 33 columnas.
- No se encontraron filas completamente duplicadas.
- La columna `BPM` presentaba 107 valores nulos y algunos valores extremos poco plausibles.
- Se identificaron participantes que declaraban escuchar música más de 12 horas al día.
- Las cuatro variables de salud mental no presentaban valores nulos.

### Decisiones de limpieza

Se eliminó la columna `BPM`, ya que no era necesaria para las hipótesis y presentaba el problema anteriormente mencionado.

También se transformaron las frecuencias de escucha musical en valores numéricos ordenados:

| Frecuencia | Valor |
|---|---|
| Never | 0 |
| Rarely | 1 |
| Sometimes | 2 |
| Very frequently | 3 |

Se conservaron los registros originales y se decidió tratar los valores nulos según las necesidades de cada análisis, evitando eliminar participantes innecesariamente.

Después de la preparación, el dataset contiene **736 registros y 32 columnas**.

La limpieza y transformación se encuentran documentadas en `02_limpieza_preparacion.ipynb` y reproducidas mediante la función `preparar_datos()` del ejecutable.

## 4. Resultados del análisis exploratorio

Una vez preparados los datos, se analizaron las cuatro hipótesis planteadas inicialmente.

Para ello se utilizaron técnicas de análisis descriptivo con Pandas y visualizaciones con Matplotlib y Seaborn.

A continuación se presentan los principales resultados obtenidos.

### H1. Horas de escucha musical y ansiedad

Se utilizó un gráfico de dispersión para explorar la relación entre las horas diarias de escucha musical y las puntuaciones de ansiedad.

La mayor concentración de observaciones se encuentra entre 0 y 8 horas diarias. Sin embargo, las puntuaciones de ansiedad presentan una gran variabilidad incluso entre participantes con hábitos de escucha similares.

**Resultado:** no se observa una tendencia visual claramente creciente o decreciente entre las horas de escucha y los niveles de ansiedad.

### H2. Género musical favorito y depresión

Se agruparon los participantes según su género musical favorito y se calculó la puntuación media de depresión de cada grupo.

Los resultados muestran diferencias descriptivas entre géneros. Lofi y Hip hop presentan las puntuaciones medias más elevadas, mientras que Gospel, Latin y R&B muestran algunas de las más bajas.

Sin embargo, existen diferencias importantes en el número de participantes de cada género. Por ejemplo, Rock cuenta con 188 respuestas, mientras que Latin solamente tiene 3.

**Resultado:** se observan diferencias en las puntuaciones medias de depresión según el género favorito, pero estas deben interpretarse con precaución, especialmente en los grupos pequeños.

```python
display(Image(filename="codigo/graficos/H2_genero_depresion.png"))
```

### H3. Frecuencia de escucha musical e indicadores de salud mental

Para esta hipótesis se compararon las puntuaciones medias de ansiedad, depresión, insomnio y TOC entre los participantes que nunca escuchan un género musical y quienes lo escuchan muy frecuentemente.

Las diferencias se representaron mediante un mapa de calor.

Entre los resultados más destacados encontramos diferencias positivas en depresión para Rock (+1,61), Metal (+1,26) y Rap (+1,24), así como en insomnio para Metal (+1,16).

**Resultado:** se observan diferencias descriptivas relacionadas con la frecuencia de escucha de algunos géneros, aunque no existe un patrón uniforme en todos los indicadores.

Estas diferencias no representan efectos causales de la música sobre la salud mental.

```python
display(Image(filename="codigo/graficos/H3_frecuencia_salud_mental.png"))
```

### H4. Efectos percibidos de la música y salud mental

Se compararon las puntuaciones medias de los cuatro indicadores psicológicos según la percepción que tienen los participantes de los efectos de la música: `Improve`, `No effect` y `Worsen`.

El grupo `Worsen` presenta las puntuaciones medias más elevadas en los cuatro indicadores, destacando depresión (7,18) y ansiedad (6,76).

Por otro lado, el grupo `No effect` presenta las medias más bajas.

Sin embargo, existe un desequilibrio importante entre los grupos: 542 participantes en `Improve`, 169 en `No effect` y solamente 17 en `Worsen`.

**Resultado:** se observan diferencias entre los grupos, pero el reducido número de participantes de `Worsen` limita la solidez de las conclusiones y su posible generalización.

## 5. Conclusiones generales

El análisis exploratorio del dataset *Music & Mental Health Survey* ha permitido identificar diferentes patrones relacionados con los hábitos musicales y los indicadores de salud mental autodeclarados.

A partir de las cuatro hipótesis analizadas, podemos destacar los siguientes resultados:

- **Horas de escucha y ansiedad (H1):** no se aprecia una tendencia visual clara entre la cantidad de horas diarias de escucha musical y los niveles de ansiedad.

- **Género favorito y depresión (H2):** se observan diferencias en las puntuaciones medias de depresión entre géneros musicales, aunque algunos grupos cuentan con pocos participantes.

- **Frecuencia de escucha y salud mental (H3):** encontramos diferencias en los indicadores psicológicos al comparar participantes que nunca escuchan determinados géneros con quienes los escuchan muy frecuentemente. Sin embargo, no existe un patrón uniforme.

- **Efectos percibidos de la música (H4):** quienes consideran que la música empeora su salud mental presentan las puntuaciones medias más elevadas en los cuatro indicadores, aunque este grupo es considerablemente más pequeño que los demás.

### Conclusión final

Los resultados muestran que existen diferencias descriptivas entre algunos hábitos musicales y las puntuaciones de salud mental, pero no permiten establecer relaciones causales.

Uno de los hallazgos más interesantes es que las personas que consideran que la música mejora su salud mental no presentan necesariamente puntuaciones bajas de ansiedad o depresión.

En definitiva, el análisis refleja la complejidad de la relación entre la música y el bienestar psicológico y demuestra la importancia de interpretar los datos teniendo en cuenta sus limitaciones.

## 6. Limitaciones del estudio

Durante el análisis se han identificado varias limitaciones que deben tenerse en cuenta:

**1. Datos autodeclarados**

Las respuestas proceden de una encuesta. Los niveles de ansiedad, depresión, insomnio y TOC son valoraciones personales y no diagnósticos clínicos.

**2. Representatividad de la muestra**

El dataset contiene 736 participantes, pero no podemos asumir que sus características representen a toda la población. Además, existe una mayor presencia de participantes jóvenes.

**3. Diferencias en el tamaño de los grupos**

Algunos géneros musicales tienen pocos participantes. Esto puede afectar a la estabilidad de sus puntuaciones medias.

Esta limitación también aparece en H4, donde solamente 17 personas consideran que la música empeora su salud mental.

**4. Valores extremos y datos ausentes**

Durante la exploración se identificaron valores extremos en las horas diarias de escucha y valores ausentes en algunas variables. Se documentaron estas características y se aplicaron las transformaciones necesarias para los análisis.

**5. Alcance del análisis**

El proyecto se centra en técnicas de análisis exploratorio y estadística descriptiva. No se han realizado pruebas para demostrar relaciones causales ni para generalizar estadísticamente los resultados a la población.

Por tanto, las conclusiones describen los patrones observados en este conjunto de datos y deben interpretarse dentro de ese contexto.
