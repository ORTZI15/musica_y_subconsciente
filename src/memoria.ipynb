# MEMORIA

## 1. Introducción y Fuentes de Datos

El proyecto **Música y Subconsciente** investiga el impacto invisible de la música comercial de alta rotación en la salud mental, el bienestar neuroemocional y la soberanía atencional del oyente, contrastando la popularidad global con métricas de valencia y energía.

### Conceptos clave
* **Música de alta rotación:** Música comercial de consumo masivo diseñada para repetirse constantemente en plataformas y listas de reproducción, optimizada para captar la atención de forma inmediata.
* **Soberanía atencional:** La capacidad que tiene una persona de controlar su propio foco mental y decidir libremente a qué presta atención, frente a estímulos externos diseñados para secuestrar su concentración.

### Fuentes de datos
Se ha utilizado el dataset principal de Spotify Tracks, compuesto inicialmente por 114.000 registros y 20 variables técnicas (`valence`, `energy`, `popularity`, `track_genre`, entre otras).

## 2. Hipótesis de Investigación

* **Hipótesis 1 (Sesgo de Valencia en el Consumo Masivo):** El ecosistema de reproducción comercial global prioriza sistemáticamente estímulos ligados a la tensión afectiva o el conflicto emocional frente a la estabilidad. Se contrastará mediante la correlación directa entre popularidad y valencia emocional.
* **Hipótesis 2 (Disonancia Acústico-Emocional en la Música Urbana):** La música urbana comercial genera disonancia cognitiva al fusionar un estímulo físico hiperactivo con un relato de pesimismo o conflicto afectivo. Se contrastará mediante análisis de densidad en el cuadrante de alta energía y baja valencia segmentado por géneros.

## 3. Limpieza y Diagnóstico Inicial

Tras la inspección estructural inicial mediante el diagnóstico de tipos de datos y valores perdidos:
* Se identificaron únicamente 3 valores nulos repartidos entre las columnas `artists`, `album_name` y `track_name`.
* Dado que el impacto sobre el volumen total es inferior al 0.003% del dataset, se procedió a su eliminación directa para preservar la integridad analítica sin alterar la distribución muestral.
* El DataFrame procesado resultante cuenta con **113.997 registros** limpios, quedando almacenado en la ruta `src/data/spotify_procesado.csv` para dar paso a la detección de duplicados y análisis de correlaciones.
